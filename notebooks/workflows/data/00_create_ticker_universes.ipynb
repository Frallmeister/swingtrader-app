# Create Ticker Universes

This notebook is used to manually create or refresh ticker universe YAML files under `src/swingtrader/configs/universes`. The generated YAML files are curated, committed artifacts that downstream runtime code reads; universe generation itself is a local/bootstrap workflow, not normal application runtime behavior.

```python
from datetime import datetime

import yaml

from swingtrader.core.paths import find_repo_root
from swingtrader.data.ingestion.universe import create_ticker_universe_file

# Anchor to the repo root so paths resolve correctly regardless of the notebook's
# working directory (the kernel cwd is this notebook's folder, not the repo root).
REPO_ROOT = find_repo_root()
UNIVERSE_CONFIG_DIR = REPO_ROOT / "src/swingtrader/configs/universes"
```

```python
import pandas as pd
names = [ticker.split(".")[0] for ticker in LARGE_CAP_TICKERS]
groups = pd.DataFrame([(name, *name.split("-")) for name in names if len(name.split("-")) > 1], columns=["ticker", "company", "kind"])
duplicated_tickers = groups.groupby("company", sort=False)["kind"].nunique().rename("nunique").to_frame().query("nunique > 1").index
duplicated_tickers
groups.query("company in @duplicated_tickers").sort_values(["company", "kind"]).query("kind in ('A', 'PREF')")["ticker"].to_list()
```

Sanity check output file.
